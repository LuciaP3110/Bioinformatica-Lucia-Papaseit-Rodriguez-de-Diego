#### 12.- Crea un archivo mediante código que se llame _tabla-nombres.txt_ y que esté formado por 4 columnas y 4 filas con la siguiente formación (1,5 ptos):
#### Primera columna: Nombre
#### Segunda columna: Apellidos (2 apellidos)

#### Cuarta columna: Mascotas (nombre de la mascota si tienen)

#### ¿Qué pasa si abres el archivo desde Jupyter y has metido acentos? ¿Puedes solucionarlo?

```python
# 12
ruta = "tabla-nombres.txt"
lineas = [
    "Nombre\tApellidos\tEdad\tMascotas",
    "Ana\tGarcía López\t24\tNiebla",
    "Luis\tMartín Pérez\t31\t",
    "Marta\tSánchez Ríos\t27\tLuna",
    "Óscar\tDíaz Núñez\t22\tPecas",
]

with open(ruta, "w", encoding="utf-8", newline="\n") as f:
    f.write("\n".join(lineas))

# Lectura de verificación (importante: encoding="utf-8")
with open(ruta, "r", encoding="utf-8") as f:
    print(f.read())
```

#### 13.- Crea un diccionario que contenga las siguientes parejas. Obtén el listado de keys() y guárdalo en una lista. Añade un par de elementos más y después convierte esa lista en un set (1 pto).

#### Rattus = novergicus, rattus
#### Glaucidium = brasilianum, hoskinsii, californicum, minutissimum
#### Echinocactus = grusonii
#### Opuntia = 1


```python
# 13
especies = {
    "Rattus": ["novergicus", "rattus"],
    "Glaucidium": ["brasilianum", "hoskinsii", "californicum", "minutissimum"],
    "Echinocactus": ["grusonii"],
    "Opuntia": [1],  # se puede mezclar tipos, pero es mejor mantener consistencia
}

# Listado de keys y guardarlo en una lista
claves = list(especies.keys())

# Añade un par de elementos más
especies["Mus"] = ["musculus"]
especies["Drosophila"] = ["melanogaster", "simulans"]

# Actualizamos la lista de claves y la convertimos en set
claves_actualizadas = list(especies.keys())
conjunto_claves = set(claves_actualizadas)

claves, claves_actualizadas, conjunto_claves
```
